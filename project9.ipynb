```python
import pandas as pd
import matplotlib.pyplot as plt

df = None
plot = None
```

```python
data = {
    "SalesID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Product": ["Product A", "Product B", "Product C", "Product D", "Product E",
                "Product A", "Product B", "Product C", "Product D", "Product E"],
    "Region": ["North", "East", "West Coast", "South", "Central",
               "North", "East", "West Coast", "South", "Central"],
    "Sales": [500, 600, 700, 800, 550, 650, 750, 850, 900, 700],
    "Year": [2022, 2022, 2022, 2022, 2022, 2023, 2023, 2023, 2023, 2023]
}

pd.DataFrame(data).to_csv("sales_data.csv", index=False)

df = None
plot = None
```

```python
def load_dataset():
    global df

    print("\n== Load Dataset ==")
    path = input("Enter the path of the dataset (CSV file): ")

    df = pd.read_csv(path)

    print("Dataset loaded successfully!")
```

```python
def explore_data():

    print("\n== Explore Data ==")
    print("1. Display the first 5 rows")
    print("2. Display the last 5 rows")
    print("3. Display column names")
    print("4. Display data types")
    print("5. Display basic info")

    choice = input("Enter your choice: ")

    if choice == "1":
        print(df.head())

    elif choice == "2":
        print(df.tail())

    elif choice == "3":
        print(df.columns)

    elif choice == "4":
        print(df.dtypes)

    elif choice == "5":
        df.info()
```

```python
def dataframe_operations():

    print("\n== Perform DataFrame Operations ==")
    print("1. Sort data")
    print("2. Filter data")
    print("3. Add column")
    print("4. Drop column")

    choice = input("Enter your choice: ")

    if choice == "1":
        column = input("Enter column name to sort: ")
        print(df.sort_values(by=column))

    elif choice == "2":
        column = input("Enter column name: ")
        value = input("Enter value: ")
        print(df[df[column].astype(str) == value])

    elif choice == "3":
        column = input("Enter new column name: ")
        value = input("Enter value: ")
        df[column] = value
        print(df)

    elif choice == "4":
        column = input("Enter column name to drop: ")
        df.drop(column, axis=1, inplace=True)
        print(df)
```

```python
def missing_data():

    print("\n== Handle Missing Data ==")
    print("1. Display rows with missing values")
    print("2. Fill missing values with mean")
    print("3. Drop rows with missing values")
    print("4. Replace missing values with a specific value")

    choice = input("Enter your choice: ")

    if choice == "1":
        missing = df[df.isnull().any(axis=1)]

        if missing.empty:
            print("\nNo missing values found in the dataset!")
        else:
            print(missing)

    elif choice == "2":
        df.fillna(df.mean(numeric_only=True), inplace=True)

    elif choice == "3":
        df.dropna(inplace=True)

    elif choice == "4":
        value = input("Enter value: ")
        df.fillna(value, inplace=True)


def statistics():

    print("\n== Descriptive Statistics ==")
    print(df.describe())
```

```python
def visualization():

    global plot

    print("\n== Data Visualization ==")
    print("1. Bar Plot")
    print("2. Line Plot")
    print("3. Scatter Plot")
    print("4. Pie Chart")
    print("5. Histogram")
    print("6. Stack Plot")

    choice = input("Enter your choice: ")

    if choice == "1":
        column = input("Enter column name: ")

        for col in df.columns:
            if col.lower() == column.lower():
                column = col

        print("Generating bar plot...")
        df[column].value_counts().plot(kind="bar")
        plot = plt.gcf()
        plt.show()
        print("Bar plot displayed successfully!")

    elif choice == "2":
        column = input("Enter column name: ")

        for col in df.columns:
            if col.lower() == column.lower():
                column = col

        print("Generating line plot...")
        plt.plot(df[column])
        plot = plt.gcf()
        plt.show()
        print("Line plot displayed successfully!")

    elif choice == "3":

        print("\n== Scatter Plot ==")

        x = input("Enter x-axis column name: ")
        y = input("Enter y-axis column name: ")

        for col in df.columns:
            if col.lower() == x.lower():
                x = col

            if col.lower() == y.lower():
                y = col

        print("Generating scatter plot...")

        plt.scatter(df[x], df[y])
        plt.xlabel(x)
        plt.ylabel(y)
        plt.title("Scatter Plot")

        plot = plt.gcf()

        plt.show()

        print("Scatter plot displayed successfully!")

    elif choice == "4":
        column = input("Enter column name: ")

        for col in df.columns:
            if col.lower() == column.lower():
                column = col

        print("Generating pie chart...")
        df[column].value_counts().plot(kind="pie")
        plot = plt.gcf()
        plt.show()
        print("Pie chart displayed successfully!")

    elif choice == "5":
        column = input("Enter column name: ")

        for col in df.columns:
            if col.lower() == column.lower():
                column = col

        print("Generating histogram...")
        df[column].plot(kind="hist")
        plot = plt.gcf()
        plt.show()
        print("Histogram displayed successfully!")

    elif choice == "6":
        column = input("Enter column name: ")

        for col in df.columns:
            if col.lower() == column.lower():
                column = col

        print("Generating stack plot...")
        df[column].plot(kind="area")
        plot = plt.gcf()
        plt.show()
        print("Stack plot displayed successfully!")
```

```python
def save_visualization():

    print("\n== Save Visualization ==")

    filename = input(
        "Enter file name to save the plot (e.g., scatter_plot.png): "
    )

    plot.savefig(filename)

    print(
        "Visualization saved as "
        + filename
        + " successfully!"
    )
```

```python
while True:

    print("\n========== Data Analysis & Visualization Program ==========")
    print("Please select an option:")
    print("1. Load Dataset")
    print("2. Explore Data")
    print("3. Perform DataFrame Operations")
    print("4. Handle Missing Data")
    print("5. Generate Descriptive Statistics")
    print("6. Data Visualization")
    print("7. Save Visualization")
    print("8. Exit")
    print("==========================================================")

    choice = input("\nEnter your choice: ")

    if choice == "1":
        load_dataset()

    elif choice == "2":
        explore_data()

    elif choice == "3":
        dataframe_operations()

    elif choice == "4":
        missing_data()

    elif choice == "5":
        statistics()

    elif choice == "6":
        visualization()

    elif choice == "7":
        save_visualization()

    elif choice == "8":
        print("\nExiting the program. Goodbye!")
        break
```
